# 1. 교재 수록 코드

```python
import matplotlib.pyplot as plt

from sklearn import datasets, metrics
from sklearn.model_selection import train_test_split

digits = datasets.load_digits()
plt.imshow(digits.images[0], cmap=plt.cm.gray_r, interpolation='nearest')


n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=6)

X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2)

knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print(scores)

plt.imshow(X_test[10].reshape(8,8), cmap=plt.cm.gray_r, interpolation='nearest')
y_pred = knn.predict([X_test[10]])
print(y_pred)
```

# 2.  Random Forest, Logistic Regression, Decision Tree

```python
# ============================================================
# Random Forest
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print("Random Forest Accuracy:", scores)

y_pred = rf.predict([X_test[10]])
print("Random Forest Prediction:", y_pred)
```

```python
# ============================================================
# Logistic Regression
# ============================================================

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=5000)

lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print("Logistic Regression Accuracy:", scores)

y_pred = lr.predict([X_test[10]])
print("Logistic Regression Prediction:", y_pred)
```

```python
# ============================================================
# Decision Tree
# ============================================================

from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()

dt.fit(X_train, y_train)

y_pred = dt.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print("Decision Tree Accuracy:", scores)

y_pred = dt.predict([X_test[10]])
print("Decision Tree Prediction:", y_pred)
```
